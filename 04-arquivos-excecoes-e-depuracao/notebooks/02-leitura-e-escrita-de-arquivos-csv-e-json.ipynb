### Modulo CSV

Arquivos CSV são usandos para representar **tabelas e dados.**  
Eles são muito comuns em:
- panilhas
- análise de dados
- exportação de sistemas  

Python possui um módulo chamado csv para facilitar o trabalho com esse formato.

```python
import csv

dados = [
    ["Nome", "Idade", "Cidade"],
    ["Alice", 30, "São Paulo"],
    ["Bob", 25, "Rio de Janeiro"],
    ["Charlie", 35, "Brasília"]
]

with open("dados.csv", "w", newline="") as arquivo_csv:
    escritor_csv = csv.writer(arquivo_csv)
    escritor_csv.writerows(dados)
```

```python
# Lendo o arquivo csv
with open("dados.csv", "r") as arquivo_csv:
    leitor_csv = csv.reader(arquivo_csv)
    for linha in leitor_csv:
        print(linha)
```

### Modulo JSON

JSON é um formato usado para armazenar e transmitir dados estruturados
Eles são muito comuns em:
- APIs
- aplicações web
- armazzenamento de dados 

Python possui um módulo chamado json para facilitar o trabalho com esse formato.

```python
import json 

dados = {
    "nome": "Alice",
    "idade": 30,
    "cidade": "São Paulo"
}

with open("dados.json", "w") as arquivo_json:
    json.dump(dados, arquivo_json)  # (informações que serão gravadas, nome_arquivo)
```

```python
# Lendo json
with open("dados.json", "r") as arquivo_json:
    dados_lidos = json.load(arquivo_json)
    print(dados_lidos)
```
