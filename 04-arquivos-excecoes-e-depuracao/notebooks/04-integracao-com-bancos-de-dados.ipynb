### Integração com banco de dados

Além de arquivos, também podemos armazenar dados em bancos de dados  
Um banco de dados permite:
- armazenar grandes quantidades de informação
- organizar dados em tabelas
- consultar dados rapidamente

Em python podemos usar vários bancos de dados  

```python
# criando ou conectando ao banco de dados SQLite

import sqlite3

conn = sqlite3.connect('../data/dadosDIO.db')

cursor = conn.cursor()

cursor.execute('''
CREATE TABLE IF NOT EXISTS usuarios (
    NAME TEXT,
    AGE INTEGER
)
''')

conn.commit() # salvando as alterações no banco de dados
```

```python
# Inserido (persistindo) informações no banco de dados
cursor.execute('''
INSERT INTO usuarios (NAME, AGE) VALUES (?, ?)''', 
('Ana', 30)
)
conn.commit() # salvando as alterações no banco de dados
```

```python
# Lendo os dados do banco de dados
cursor.execute('SELECT * FROM usuarios')

dados = cursor.fetchall()

print(dados)
```

```python
# Encerrando a conexão com o banco de dados
cursor.close() # fecha o cursor
conn.close()   # fecha a conexão e libera o arquivo dadosDIO.db

print("Conexão encerrada.")
```
