### O que é uma API ?

API significa **Application Programming Interface**  
Uma API permite que **um sistema forneça dados para outro sistema**  
Isto é, "podemos" **pedir informações para um serviço na internet** sem acessar diretamente o banco de dados ou arquivo.  
Exemplos de APIs populares:
- previsão do tempo
- dados de criptmoedas
- dados de usuários
- mapas e localização

### Leitura de dados externos (API)

Normalmente as APIs retornam dados no formato **JSON**  
Podemos converter essa resposta diretamente em um **dicionário python**

### Requisições HTTP

Para acessar uma API é necessário fazer uma **requisição HTTP**    
O tipo mais comum é:  
GET &rarr; usado para **buscar dados**  
Em python usamos a biblioteca **requests** para fazer requisições

```python
# Teste requisão HTTP

import requests

url = 'https://api.agify.io?name=Junior'

response = requests.get(url) # resposta da resquisição

print(response.status_code) # status da requisição
print(response.text) # conteúdo da resposta
```

```python
# Leitura do conteúdo da resposta em formato JSON
data = response.json() # conteúdo da resposta em formato JSON
print(data) # exibe os dados em formato JSON
```

```python
# Pegando os dados da API (arquivo JSON)
print('Nome: ', data['name'])
print('Idade: ', data['age'])
print('Número deregistros: ', data['count'])
```

```python
# Salvado os dados da API
import json

with open('dados_api.json', 'w') as arquivo:
    json.dump(data, arquivo) # salvando os dados da API em um arquivo JSON
    
```
