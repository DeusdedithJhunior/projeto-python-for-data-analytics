### Leitura de arquivos

Para ler arquivos usamos o modo "r" (read)  
Isso permite acessar o contéudo do arquivo sem modificá-lo

```python
from pathlib import Path
PASTA_DATA = Path("../data")
caminho = PASTA_DATA / "dados.txt"

#Criando um arquivo para texte
with open(caminho, "w") as arquivo:
    arquivo.write("Python é uma linguagem poderosa. \n")
    arquivo.write("Estamos aprendendo arquivos.")

print(caminho.resolve()) # mostra o caminho completo do arquivo
```

```python
# Lendo o arquivo completo
with open(caminho, "r") as arquivo:
    conteudo = arquivo.read()
    print(conteudo)
```

```python
# Lendo o arquivo linha por linha (iterando sobre o arquivo)
with open(caminho, "r") as arquivo:
    for linha in arquivo:
        print(linha.strip())  # strip() remove espaços em branco e quebras de linha
```

### Gravação de arquivos

Para gravar informações em um arquivo usamos o modo "w" (write).  
Esse modo cria o arquivo caso ela não exista.  
Se o arquivo já existir, ele será sobrescrito (apagando o conteúdo anterior).

```python
# Escrevendo em um arquivo
caminho_escrita = PASTA_DATA / "relatorio.txt"
with open(caminho_escrita, "w") as arquivo:
    arquivo.write("Relatório de vendas\n")
    arquivo.write("Produto A: 100 unidades\n")
    arquivo.write("Produto B: 200 unidades\n")
    arquivo.write("Total: 300 unidades\n")
```

```python
# Adicionando conteúdo a um arquivo existente
# usando o modo "a" (append)
caminho_append = PASTA_DATA / "relatorio.txt"
with open(caminho_append, "a") as arquivo:
    arquivo.write("Produto C: 150 unidades\n")
    arquivo.write("Produto D: 250 unidades\n")
    arquivo.write("Total atualizado: 700 unidades\n")
```

```python
# Lendo o arquivo linha por linha
with open(caminho_append, "r") as arquivo:
    for linha in arquivo:
        print(linha.strip())
```
